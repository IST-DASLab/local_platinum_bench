```python
%env CUDA_VISIBLE_DEVICES=8
%load_ext autoreload
%autoreload 2
import platinumbench
from types import SimpleNamespace
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python

model_id = "meta-llama/Llama-3.2-1B-Instruct"  # HF repo

# choose device
device = "cuda" if torch.cuda.is_available() else "cpu"

# tokenizer + model
tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=False)
model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16 if device=="cuda" else torch.float32)
model.to(device)
```

```python
args =SimpleNamespace()
args.model = model
args.tokenizer=tokenizer
model_name="LLama3.2-1B-Instruct"
platinumbench.run_benchmark(model_list={model_name:(args.model,args.tokenizer)},output_file="./outputs/results.csv", dataset_names=["singleop","singleq"],seed=43)
```
